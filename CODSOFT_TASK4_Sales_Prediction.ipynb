# Task 4 - Sales Prediction Using Python

### Objective

To build a machine learning model that predicts product sales based on advertising expenditure.

```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("Libraries imported successfully!")
```

```python
# Create a sample advertising dataset for the project

data = {
    'TV': [230.1, 44.5, 17.2, 151.5, 180.8, 8.7, 57.5, 120.2, 8.6, 199.8],
    'Radio': [37.8, 39.3, 45.9, 41.3, 10.8, 48.9, 32.8, 19.6, 2.1, 49.8],
    'Newspaper': [69.2, 45.1, 69.3, 58.5, 58.4, 75.0, 23.5, 11.6, 1.0, 21.2],
    'Sales': [22.1, 10.4, 9.3, 18.5, 12.9, 7.2, 11.8, 13.2, 4.8, 19.6]
}

df = pd.DataFrame(data)

print("Sales dataset loaded successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

## 1. Understanding the Dataset

The dataset contains advertising expenditure and corresponding sales values.

We will use these values to understand the relationship between advertising and sales.

```python
# Display basic information

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset:")
print(df)
```

## 2. Data Visualization

Visualization helps us understand the relationship between advertising expenditure and sales.

```python
# Visualize TV advertising and sales

plt.figure(figsize=(7,5))

plt.scatter(df['TV'], df['Sales'])

plt.title('TV Advertising vs Sales')
plt.xlabel('TV Advertising')
plt.ylabel('Sales')

plt.show()
```

## 3. Preparing Data for Machine Learning

TV, Radio and Newspaper advertising expenses will be used as input features, while Sales will be the target variable.

```python
# Define features and target

X = df[['TV', 'Radio', 'Newspaper']]
y = df['Sales']

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
# Split data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

## 4. Machine Learning Model

A Linear Regression model is used to predict sales based on advertising expenditure.

```python
# Create and train Linear Regression model

model = LinearRegression()

model.fit(X_train, y_train)

print("Model trained successfully!")
```

## 5. Model Prediction and Evaluation

The trained model is used to predict sales on the test data. The predictions are then compared with the actual sales values.

```python
# Make predictions

y_pred = model.predict(X_test)

print("Predicted Sales:")
print(y_pred)

# Evaluate the model

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("\nMean Absolute Error:", round(mae, 2))
print("Root Mean Squared Error:", round(rmse, 2))
print("R2 Score:", round(r2, 2))
```

## 6. Actual vs Predicted Sales

The following graph compares the actual sales values with the values predicted by the machine learning model.

```python
# Compare actual and predicted sales

plt.figure(figsize=(8,5))

plt.plot(y_test.values, label='Actual Sales', marker='o')
plt.plot(y_pred, label='Predicted Sales', marker='x')

plt.title('Actual vs Predicted Sales')
plt.xlabel('Test Sample')
plt.ylabel('Sales')

plt.legend()
plt.show()
```

A Linear Regression model was trained using TV, Radio and Newspaper advertising expenditure to predict sales.

The model was evaluated using Mean Absolute Error, Root Mean Squared Error and R2 Score.
